```python
while True:

    print("\n--- SMART HOME MENU ---")
    print("1. Light Control")
    print("2. Fan Control")
    print("3. Motion Detection")
    print("4. Play Songs")
    print("5. Stop Songs")
    print("6. Robot Vacuum Cleaner")
    print("7. Stop System")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        light = (input("Is the room visible? (yes/no)"))

        if light == "no":
            print("Turn ON the Light")
        else:
            print("Turn OFF the Light")

    elif choice == 2:
        temperature = int(input("Enter temperature: "))

        if temperature > 30:
            print("Fan is ON")
        else:
            print("Fan is OFF")

    elif choice == 3:
        motion = input("Is motion detected? (yes/no): ")

        if motion == "yes":
            print("Alarm is ON")
        else:
            print("Alarm is OFF")

    elif choice == 4:
        print("Songs are playing")

    elif choice == 5:
        print("Songs are stopped")

    elif choice == 6:
        floor = input("Is the floor dirty? (yes/no): ")

        if floor == "yes":
            print("Robot vacuum cleaner started cleaning")
        else:
            print("Floor is clean. Vacuum cleaner is OFF")

    elif choice == 7:
        print("Smart home system stopped")
        break

    else:
        print("Invalid choice. Please enter again.")
```
