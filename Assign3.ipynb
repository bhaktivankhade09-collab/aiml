```python
import heapq as queue

def gbfs(graph,h,start,goal):
    openlist=[]
    closelist=[]
    parent={start:None}
    
    queue.heappush(openlist,(h[start],start))
    
    while openlist:
        minnode=queue.heappop(openlist)
        node=minnode[1]
        print("Min Node : ",node)
        closelist.append(node)
        print("Closedlist : ",closelist)
        
        if node==goal:
            path=[]
            while node:
                path.append(node)
                node=parent[node]
            path.reverse()
            print("Path : ",path)
            print("Found")
            return
            
            
        for m in graph[node]:
            if m not in closelist:
                parent[m]=node
                queue.heappush(openlist,(h[m],m))
        print("OpenList : ",openlist)
    print("Goal not found")           
            

graph={}
h={}
n=int(input("Enter Number Of Vertices"))
for i in range(n):
    node=input("Enter vertex:")
    graph[node]=list((input("Enter Child Vertex:")))
    h[node]=int(input("Enter heuristic Value : "))
print(graph)
print(h)
start=input("Enter Start of the Graph:")
goal=input("Enter Goal:")
gbfs(graph,h,start,goal)
```
