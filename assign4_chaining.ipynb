```python
def forward_chain(kb, facts):
    known_facts = facts.copy()
    new_added_fact = True

    while new_added_fact:
        new_added_fact = False

        for rule in kb:
            antecedents, consequent = rule

            if consequent not in known_facts:
                if set(antecedents).issubset(known_facts):
                    known_facts.append(consequent)
                    new_added_fact = True

    return known_facts


def backward_chain(kb, facts, goal, steps=None):
    if steps is None:
        steps = []

    if goal in facts:
        steps.append("Fact found: " + goal)
        return True

    for rule in kb:
        antecedents, consequent = rule

        if consequent == goal:
            all_true = True

            for fact in antecedents:
                if not backward_chain(kb, facts, fact, steps):
                    all_true = False
                    break

            if all_true:
                steps.append("Rule applied: " + " AND ".join(antecedents) + " -> " + consequent)
                return True

    return False


ruleset = [
    (["battery_low", "dim_lights"], "alternator_fault"),
    (["alternator_fault"], "battery_not_charging"),
    (["battery_not_charging"], "replace_alternator")
]


while True:
    print("1. Forward Chaining")
    print("2. Backward Chaining")
    print("3. Exit")

    choice = input("Enter your choice: ")
    
    knownfacts = ["battery_low", "dim_lights"]

    if choice == "1":

        print("\nForward Chaining")
        print("Initial Facts:")

        for fact in knownfacts:
            print(fact)

        result = forward_chain(ruleset, knownfacts)

        print("\nDerived Facts:")

        for fact in result:
            print(fact)

    elif choice == "2":

        print("\nBackward Chaining")

        print("Initial Facts:")
        for fact in knownfacts:
            print(fact)

        goal = input("\nEnter the goal to check: ")

        steps = []

        if backward_chain(ruleset, knownfacts, goal, steps):
            print("\nGoal:", goal)
            print("\nproved")

            for step in reversed(steps):
                print(step)

        else:
            print("\nGoal:", goal)
            print("Result: Goal cannot be proved.")

    elif choice == "3":

        print("\nExit")
        break

    else:
        print("\nInvalid choice")
```
